```python
  # 1 comando "input' para receber ao menos 2 números de entrada do usuário;

num1 = input("Digite o primeiro numero")
num2 = input('Digite o segundo numero')

#2 - Converta os valores recebidos pelo usuário para númrointeiro (int) pu ponto flutuante n(float);

num1 = float(num1)
num2 = float(num2)

#3 - Implemente ao menos 4 operações matemáticas em seu código;

soma = num1 + num2
subtracao = num1 - num2
multiplicacao = num1 * num2
divisao = num1 / num2


#4 - Adicione um laço de repetição ou uma condicional. Por exemplo você pode

operação = input("Dígito o número da operação desejada: 1 -soma 2- subtração, 3 - multiplicação ou 4 -divisão")


if  operação == "1":
   print('Resultado :', soma)
elif operação == "2":
   print('Resultado :', subtracao)
elif  operação  == "3":
   print('Resultado :', divisao)
elif operação == '4':
    print('Resultado :', multiplicacao)
else:
   print('Resultado!', divisao)
   print("Operacao invalida!")
```
