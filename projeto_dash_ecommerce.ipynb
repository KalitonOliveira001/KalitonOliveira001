```python
import dash
from dash import html, dcc
import pandas as pd
import plotly.express as px
```

```python

# Lê o CSV carregar os dados
df = pd.read_csv("ecommerce_estatistica.csv")
```

```python
['Unnamed: 0', 'Título', 'Nota', 'N_Avaliações', 'Desconto', 'Marca', 'Material',
 'Gênero', 'Temporada', 'Review1', 'Review2', 'Review3', 'Qtd_Vendidos', 'Preço',
 'Nota_MinMax', 'N_Avaliações_MinMax', 'Desconto_MinMax', 'Preço_MinMax',
 'Marca_Cod', 'Material_Cod', 'Temporada_Cod', 'Qtd_Vendidos_Cod', 'Marca_Freq', 'Material_Freq']
```

```python
grafico_vendas = px.bar(df, x="Marca", y="Qtd_Vendidos", title="Quantidade Vendida por Marca")
```

```python
grafico_lucro = px.line(df, x="Preço", y="Nota", title="Nota Média por Faixa de Preço", markers=True)
```

```python

# Gráfico 3: Distribuição de Descontos
grafico_descontos = px.histogram(df, x="Desconto", title="Distribuição de Descontos (%)")
```

```python

# Inicializa o app

# Inicializa o app
app = dash.Dash(__name__)
app.title = "Dashboard E-commerce"

```

```python

# Layout do Dash
app.layout = html.Div([
    html.H1("Dashboard de Estatísticas de E-commerce", style={'textAlign': 'center'}),

    html.Div([
        html.H2("Gráfico de Vendas"),
        dcc.Graph(figure=grafico_vendas)
    ]),

    html.Div([
        html.H2("Gráfico de Lucro"),
        dcc.Graph(figure=grafico_lucro)
    ])
])
```

```python
#
if __name__ == '__main__':
    app.run(debug=True)
```
