```python
# Criar gráficos para análise dos dados
# Gráfico de Histograma
plt.figure(figsize=(10, 6))
sns.histplot(df['Preço'], kde=True)  # Substitua 'Preço' por outra coluna se quiser
plt.title('Histograma de Preços')
plt.xlabel('Preço')
plt.ylabel('Frequência')
plt.show()
```

```python
# Gráfico de Dispersão
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df['Preço'], y=df['Qtd_Vendidos'])
plt.title('Relação entre Preço e Quantidade Vendida')
plt.xlabel('Preço do Produto')
plt.ylabel('Quantidade Vendida')
plt.show()
```

```python

# Mapa de Calor
plt.figure(figsize=(12, 6))

# Selecionar apenas colunas numéricas para calcular a correlação
df_numerico = df.select_dtypes(include=['number'])

sns.heatmap(df_numerico.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Mapa de Calor das Correlações')
plt.show()
```

```python

import pandas as pd

# Carregar o arquivo CSV
df = pd.read_csv('ecommerce_estatistica.csv')
```

```python
#Gráfico de Barras
plt.figure(figsize=(12, 6))
df.groupby('Marca')['Preço'].mean().sort_values().plot(kind='bar', color='blue')
plt.title('Preço Médio por Marca')
plt.xlabel('Marca')
plt.ylabel('Preço Médio')
plt.xticks(rotation=90)  # Rotaciona os nomes das marcas
plt.show()
```

```python
#Gráfico de Pizza
plt.figure(figsize=(8, 8))
df['Temporada'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90, colormap='Pastel1')
plt.title('Distribuição de Vendas por Temporada')
plt.ylabel('')  # Remove o rótulo do eixo Y para melhorar visualização
plt.show()

```

```python
# Gráfico de Densidade
plt.figure(figsize=(10, 6))
sns.kdeplot(df['Preço'], shade=True)
plt.title('Distribuição de Densidade dos Preços')
plt.xlabel('Preço')
plt.ylabel('Densidade')
plt.show()
```

```python
import pandas as pd
```

```python
# Carregar o arquivo CSV
df = pd.read_csv('ecommerce_estatistica.csv')
```

```python
print(df.head())
```

```python
# Certifique-se de que as colunas são numéricas
df['Preço'] = pd.to_numeric(df['Preço'], errors='coerce')
df['Qtd_Vendidos'] = pd.to_numeric(df['Qtd_Vendidos'], errors='coerce')
```

```python

# Remover valores nulos para evitar erros no gráfico
df_clean = df.dropna(subset=['Preço', 'Qtd_Vendidos'])
```

```python
# Criar o gráfico de regressão
plt.figure(figsize=(10, 6))
sns.regplot(x=df_clean['Preço'], y=df_clean['Qtd_Vendidos'])
plt.title('Regressão entre Preço e Quantidade Vendida')
plt.xlabel('Preço do Produto')
plt.ylabel('Quantidade Vendida')
plt.show()
```
