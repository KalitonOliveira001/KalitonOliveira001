ANÁLISE EXPLORATÓRIA (EDA)

```python
# Converter data para datetime
df['date'] = pd.to_datetime(df['date'])
```

```python
# Filtrar só o Brasil
brasil = df[df['location'] == 'Brasil']
```

```python
# Verificar valores ausentes
brasil.isna().sum()
```

 Criar colunas úteis

```python
# Filtrar novamente só os dados do Brasil
df_brasil = df[df['location'] == 'Brazil'].copy()

# Criar colunas úteis
df_brasil['date'] = pd.to_datetime(df_brasil['date'])
df_brasil['casos_100mil'] = (df_brasil['total_cases'] / df_brasil['population']) * 100000
df_brasil['mortes_100mil'] = (df_brasil['total_deaths'] / df_brasil['population']) * 100000

# Selecionar apenas as colunas essenciais
colunas = ['date', 'total_cases', 'new_cases', 'total_deaths', 'new_deaths',
           'people_vaccinated', 'population', 'casos_100mil', 'mortes_100mil']
df_export = df_brasil[colunas]

# Ver as 5 primeiras linhas antes de exportar
df_export.head()
```

```python
# Salva como CSV menor (com no máximo 1 país e colunas úteis)
df_export.to_csv('covid_brasil_tratado.csv', index=False)
```
